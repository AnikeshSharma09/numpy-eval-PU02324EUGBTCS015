```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
A = np.random.randint(1, 11, size=(4, 3))

print("Matrix A:")
print(A)
print("Shape of A:", A.shape)
```

```python
B = np.random.randint(1, 11, size=(3, 5))

print("Matrix B:")
print(B)
print("Shape of B:", B.shape)
```

```python
assert A.shape[1] == B.shape[0], "Inner dimensions do not match"

print("Dimension validation successful.")
```

```python
C = np.dot(A, B)

print("Matrix C = A · B:")
print(C)
print("Shape of C:", C.shape)
```

```python
print("Total sum of C:", np.sum(C))
```

```python
print("Row-wise sums:", np.sum(C, axis=1))
```
